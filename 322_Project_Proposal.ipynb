## Dataset Description

https://github.com/vaastav/Fantasy-Premier-League

https://github.com/olbauday/FPL-Core-Insights

The two datasets include data recorded for Fantasy Premier League (FPL). The vaastav data includes data from each match of each game week (GW) from the 2016-17 season to the present. The olbauday data includes the same, although it only has data from the 2024-25 season onwards. Each dataset has aggregated data for players and teams. The vaastav data specifically includes fantasy data from the 2026 World Cup and combined data from every season.

We will be trying to predict whether a player is classified as a defender, midfielder, or forward. We will be using attributes such as expected goals, expected assists, goals, assists, defensive contributions, corners, free kicks, penalties, tackles, clearances, blocks, influence, etc.

## Implementation / Technical Merit

We plan to use three classifiers: our own random forest, k-NN, and Naive Bayes. We will compare them to a single decision tree and to a baseline that always guesses the most common position.

Some challenges we expect:

- A player shows up in a lot of rows (every gameweek and every season), so we have to split by player so the same player isn't in both the training and test sets. If we don't, the model could just recognize the player and our accuracy would look better than it is.
- There are fewer forwards than midfielders and defenders, so the classes are imbalanced. We will use a stratified split and look at results for each position, not just overall accuracy.
- Goalkeepers aren't one of our classes, so we will remove them.
- Players who barely played will have mostly zeros for their stats, so we will filter out players with very few minutes and probably use per-90 numbers instead of totals.
- The two datasets have different columns and some seasons are missing columns, so if we combine them we will have to line them up. We will decide which dataset (or both) to use after looking at them more.
- We expect midfielders and forwards to be the hardest to tell apart since a lot of their stats overlap.

We have a lot of attributes (expected goals, tackles, clearances, blocks, etc.), so we will use EDA like box plots of each stat by position to pick the ones that actually separate the positions. We will drop attributes that are basically duplicates (like goals and expected goals). Then we will compare accuracy with all attributes against accuracy with just the selected ones.

## Potential Impact

This project will help non-soccer players understand the roles that different positions play, and could potentially aid fantasy managers in deciding which players to prioritize based on specific strengths instead of simply name-value. Therefore, the results will have an educational purpose for casual fans and fans who are hoping to be more competitive in their FPL group.

The stakeholders are soccer fans, Premier League enthusiasts, and those involved in betting markets.
